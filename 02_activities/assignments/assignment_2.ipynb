#### Starter Code for Question 1

```python
from collections import deque
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right
       
```

```python
from collections import deque

def is_duplicate(root: TreeNode) -> int:
    if root:
        visited = {tuple(root.val) if isinstance(root.val, list) else root.val}
        queue = deque([root])
        
        while queue:
            node = queue.popleft() # to find closest duplicate to root
            for child in (node.left, node.right):
                if child:
                    val = tuple(child.val) if isinstance(child.val, list) else child.val
                    if val in visited:
                        return val
                    visited.add(val)
                    queue.append(child)
    
    return -1
```

```python
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(2)
root.left.left = TreeNode(3)
root.left.right = TreeNode(5)
root.right.left = TreeNode(6)
root.right.right = TreeNode(7)
```

```python
is_duplicate(root)
```

```python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    # TODO

    # Collect all non-zero elements
    result = [num for num in nums if num != 0]

    # Count the number of zeros
    zero_count = len(nums) - len(result)

    # Append zeros to the end
    result.extend([0] * zero_count)

    return result
```
